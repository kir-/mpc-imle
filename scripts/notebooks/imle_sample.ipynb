### Setup

```python
import mpc_imle.utils as utils
```

### Conditional sampling

```python
class Args:
  loadpath = 'logs/walker2d-medium-v2/imle/defaults_H32_S1_SF4_RWFalse'
  diffusion_epoch = 'latest'
  n_samples = 4
  device = 'cpu'
  # device = 'cuda:0'
    
args = Args()
```

Load diffusion model from disk

```python
%cd ..

diffusion_experiment = utils.load_experiment(args.loadpath, epoch=args.diffusion_epoch, generator='imle_config.pkl')

dataset = diffusion_experiment.dataset
renderer = diffusion_experiment.renderer
model = diffusion_experiment.trainer.ema_model
```

Initialize environment

```python
env = dataset.env
obs = env.reset()
```

Run imle

```python
observations = utils.colab.run_imle(model, dataset, obs, args.n_samples, args.device)
print(observations.shape)
```

Render the samples

```python
utils.colab.show_sample(renderer, observations, filename='sample.mp4', savebase='./videos')
```
